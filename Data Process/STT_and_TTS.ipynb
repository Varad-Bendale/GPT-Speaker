```python
class Head(nn.Module):
    def __init__(self, head_quant):
        super().__init__()
        self.query = nn.Linear(dim, head_quant, bias=False)
        self.key   = nn.Linear(dim, head_quant, bias=False)
        self.value = nn.Linear(dim, head_quant, bias=False)
        self.mask  = torch.tril(torch.ones(block_size, block_size))

    def forward(self, x):
        q = self.query(x)
        k = self.key(x)
        v = self.value(x)
        T = x.shape[1]
        wieghts = q @ k.transpose(-2,-1) * (head_quant ** -0.5)
        wieghts = wieghts.masked_fill(self.mask[:T, :T].to(x.device) == 0, float('-inf'))
        wieghts = torch.softmax(wieghts, dim=-1)
        pos_pattern = wieghts @ v
        return pos_pattern

class head_concat(nn.Module):
    def __init__(self, num_heads, head_quant):
        super().__init__()
        self.heads = nn.ModuleList([Head(head_quant) for _ in range(num_heads)])
        self.proj  = nn.Linear(num_heads * head_quant, dim)

    def forward(self, x):
        pos_pattern = torch.cat([head(x) for head in self.heads], dim=-1)
        return self.proj(pos_pattern)

class block_in_transformer(nn.Module) :

    def __init__(self) :
         super().__init__()
         self.head = head_concat(num_heads, head_quant)
         self.hidden_layer_1 = nn.Linear(dim , 4*dim).to(device)
         self.hidden_layer_2  = nn.Linear(4*dim , dim).to(device)
         self.norm_1 = nn.LayerNorm(dim)
         self.norm_2 = nn.LayerNorm(dim)


    def forward(self , x) :
       pattern = self.head(x)
       x = x + pattern
       x  = self.norm_1(x)
       ffn = torch.nn.functional.leaky_relu(self.hidden_layer_1(x), negative_slope=0.01)
       ffn = self.hidden_layer_2(ffn)
       x = x + ffn
       x = self.norm_2(x)
       return x
```

```python
class TTS_transformer(nn.Module):

      def __init__(self) :
          super().__init__()
          self.device = 'cuda' if torch.cuda.is_available() else 'cpu'
          self.dim        = 256
          self.n_layers   = 4
          self.num_heads  = 4
          self.head_quant = 64
          self.block_size = 128
          self.info = info
          self.vocab_size = len(self.info)
          self.batch_size = 64
          self.Blocks = nn.ModuleList([block_in_transformer() for _ in range(self.n_layers)]).to(self.device)
          self.positions  = nn.Embedding(self.block_size, 256).to(self.device)
          self.g = torch.Generator()
          self.g = self.g.manual_seed(42)
          self.cluch =  0
          self.vqvae = VQVAE()


      def defining_part(self , info ):
          self.info = info
          self.vocab_size = len(self.info)
          self.Encods     = nn.Embedding(self.vocab_size, self.dim).to(self.device)
          self.final      = nn.Linear(self.dim, self.vocab_size).to(self.device)
          token_ids = torch.tensor(list(self.info.keys()))
          q_tokens = token_ids[:-1]
          t_tokens = token_ids[1:]
          self.xs = []
          self.ys = []
          for i in range(0, len(q_tokens) - self.block_size):
              self.xs.append(q_tokens[i : i + self.block_size])
              self.ys.append(t_tokens[i : i + self.block_size])
          self.xs = torch.stack(xs)
          self.ys = torch.stack(ys)
          self.clutch = 0

      def training(self) :
          self.parameters = list(self.Encods.parameters()) + list(self.positions.parameters()) + list(self.Blocks.parameters()) + list(self.final.parameters())
          self.optimizer = torch.optim.AdamW(self.parameters, lr=1e-4)
          for i in range(10000):
            ix  = torch.randint(0, self.xs.shape[0], (self.batch_size,))
            x_  = self.xs[ix].to(self.device)
            y_  = self.ys[ix].to(self.device)

            tok_emb = self.Encods(x_)
            if self.clutch == 0  :
                pos_emb = self.positions(torch.arange(self.block_size).to(self.device))
            else :
                size = x_.shape[1]
                pos_emb = self.positions(torch.arange(size).to(self.device))
            x = tok_emb + pos_emb
            for block in self.Blocks:
                x = block(x)
            logits = self.final(x)
            loss = torch.nn.functional.cross_entropy(logits.view(-1, self.vocab_size),y_.view(-1))
            self.optimizer.zero_grad()
            loss.backward()
            self.optimizer.step()
            if i % 500 == 0:
                print(f'step {i} loss: {loss.item():.4f}')
            if loss.item() < 1.3:
                print(f'stopping at step {i}')
                break

     def alphabet_to_word_defining_part(self, info , vocab_size ):
            self.vocab_size = vocab_size
            self.Encods = nn.Embedding(self.vocab_size, self.dim).to(self.device)
            self.final = nn.Linear(self.dim, self.vocab_size).to(self.device)
            self.xs = []
            self.ys = []

            for info in infos:
                xs_tensor = torch.cat(info[0], dim=0)
                ys_tensor = info[1].flatten()
                self.xs.append(xs_tensor)
                self.ys.append(ys_tensor)
            self.xs = pad_sequence(self.xs, batch_first=True, padding_value=0)
            self.ys = pad_sequence(self.ys, batch_first=True, padding_value=0)
            self.clutch = 1

      def decode(self, sound):
          self.vqvae.get_tokens(sound)
          x_ = torch.stack(self.x_part[-1:]).to(self.device)
          tok_emb = self.Encods(x_)
          if self.clutch == 0  :
              pos_emb = self.positions(torch.arange(self.block_size).to(self.device))
          else :
              size = x_.shape[1]
              pos_emb = self.positions(torch.arange(size).to(self.device))
          x = tok_emb + pos_emb
          for block in self.Blocks:
              x = block(x)
          logits  = self.final(x)
          predicted_word = torch.argmax(logits, dim=-1)
          return predicted_word

      def get_as_personlity(self, sound):
          self.vqvae.get_tokens(sound)
          x_ = torch.stack(self.x_part[-1:]).to(self.device)
          tok_emb = self.Encods(x_)
          if self.clutch == 0  :
              pos_emb = self.positions(torch.arange(self.block_size).to(self.device))
          else :
              size = x_.shape[1]
              pos_emb = self.positions(torch.arange(size).to(self.device))
          x = tok_emb + pos_emb
          for block in self.Blocks:
              x = block(x)
          logits  = self.final(x)
          predicted_word = torch.argmax(logits, dim=-1)
          return predicted_word


      def save_alphabets_to_words():
          torch.save({'Encods'  : self.Encods.state_dict(),'positions': self.positions.state_dict(),'blocks': self.Blocks.state_dict(),'final': self.final.state_dict()}, 'transformer.pt')
          files.download('transformer_alphabets_tokens_to_words_tokens.pt')


      def save_aplhabets():
          torch.save({'Encods'  : self.Encods.state_dict(),'positions': self.positions.state_dict(),'blocks': self.Blocks.state_dict(),'final': self.final.state_dict()}, 'transformer.pt')
          files.download('transformer_alphabets_to_sound.pt')
```
